# 03. Business Fraud Analytics & Patterns

## Project: Real-Time Credit Card Fraud Detection & Analytics System
**Objective**: Conduct deep-dive business analysis into fraud behavior across transaction size cohorts, time of day, and calculate aggregate financial exposure.

```python
import os
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

ROOT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from src.data.loader import load_transaction_data
from src.analytics.fraud_analytics import (
    calculate_executive_kpis,
    get_hourly_fraud_patterns,
    get_amount_bucket_analysis,
    get_top_fraud_transactions
)

df = load_transaction_data(source='auto', clean=True)
kpis = calculate_executive_kpis(df)
for k, v in kpis.items():
    print(f"{k:<25}: {v}")
```

### 1. Cohort Analysis: Fraud Rate by Amount Bracket

```python
amt_cohorts = get_amount_bucket_analysis(df)
print(amt_cohorts.to_string(index=False))

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=amt_cohorts, x='tier', y='fraud_rate_pct', palette='Reds_d', ax=ax)
ax.set_title('Fraud Incidence Rate (%) across Transaction Amount Brackets', fontsize=13)
ax.set_ylabel('Fraud Rate (%)')
ax.set_xlabel('Amount Tier')
plt.tight_layout()
plt.show()
```

### 2. Hourly Fraud Velocity & Peak Activity Windows

```python
hourly = get_hourly_fraud_patterns(df)
top_risk_hours = hourly.sort_values(by='fraud_rate_pct', ascending=False).head(5)
print("Top 5 Peak Risk Hours:")
print(top_risk_hours[['hour', 'total_transactions', 'fraud_transactions', 'fraud_rate_pct']].to_string(index=False))
```

### 3. Highest Financial Value Fraud Incidents

```python
top_frauds = get_top_fraud_transactions(df, top_n=10)
print(top_frauds.to_string(index=False))
```
